# **2. Séries Temporais**

Uma série temporal é um conjunto de observações de uma variável ordenadas no tempo. Essas observações são coletadas em intervalos regulares, como dias, meses, trimestres ou anos, permitindo analisar a evolução dessa variável ao longo do tempo. Diferentemente de quando trabalhamos com dados do tipo cross-section (uma amostra de dados coletadas em um momento específico do tempo), as séries temporais possuem uma **ordenação natural** que é o tempo. Cada observação está associada a um momento específico do tempo e carrega informação sobre a evolução da série.

É fundamental termos em mente alguns conceitos de séries temporais quando estamos trabalhando com macroeconomia. Quando pensamos em juros, inflação, balança comercial, entre outros dados econômicos estamos sempre falando de séries temporais. Pense nas notícias de jornais: IPCA fecha o mês de janeiro em x%, PIB do Brasil cresce y% em 2023, etc...

Perceba que todas essas notícias sempre apresentam o período a que o dado se refere. Não faz sentido falarmos só que a inflação foi de 4.5%, é **fundamental** sabermos do período que o dado se refere.

📝 **Observação**: Neste case, traremos alguns tópicos que sejam mais relevantes e que com uma abordagem mais direcionada. Porém, para um estudo mais aprofundado e formal sobre séries temporais, recomendamos ver o **case de finanças quantitativas**

## **Variável aleatória**
De maneira simples, podemos pensar em uma variável aleatória como um número que representa a realização de um determinado evento que possuí alguma incerteza associada. Pense no tempo que você leva para se deslocar ao trabalho. Esse tempo não é fixo, ele na realidade possui alta incerteza devido ao trânsito, chuva, etc. Esse tempo medido é uma variável aleatória.


## **Processos estocásticos**

Um processo estocástico é basicamente um conjunto de variáveis aleatórias organizadas no tempo. Ou seja, é um fenômeno aleatório que evolui ao longo do tempo.

Suponha que agora, você mede o tempo de deslocamento ao trabalho todos os dias por um ano inteiro. Você terá uma sequência de números diferentes ao longo do tempo, mas seguindo um certo padrão (por exemplo, pode ser maior em dias de chuva ou menor em feriados). Isso é fundamental pois, apesar de cada dia haver certa incerteza associada, é possível modelar o comportamento do deslocamento ao trabalho e extrair informações extramemente úteis.

💡**Dica**: Dá para pensar em um processo estocástico como uma “história” contada por várias variáveis aleatórias ao longo do tempo.

**Definição formal de Série Temporal**:

Seja $(\Omega, \mathcal{F}, P)$ um espaço de probabilidade e seja $\{X_{t}\}_{t \in T}$ um processo estocástico, ou seja, uma família de variáveis aleatórias $X_{t}: \Omega \rightarrow \mathbb{R}$ indexadas por um conjunto $T$, que representa o tempo.

**Uma série temporal é uma realização desse processo estocástico**, isto é, uma sequência de valores observados:

$$
        \{x_{t}\}_{t \in T} \subset \mathbb{R}
$$

onde cada $x_{t}$ corresponde ao valor assumido pelo processo no instante $t$.

## **Componentes de Séries Temporais**

Os componentes de uma série temporal são os diferentes elementos que influenciam a variação dos dados ao longo do tempo. Eles ajudam a decompor uma série temporal em partes distintas para melhor análise e modelagem. Quando pensamos nos componentes das séries temporais, podemos pensar em questões do tipo: Oque está acontecendo com essa série no longo prazo? Existe algum padrão que se repete em intervalos regulares de tempo? Existe algo imprevisível que acontece com essa série?

Ao decompormos uma série temporal, temos uma maior capacidade para responder questões desse tipo. Os principais componentes são:

### **Tendência**

A tendência representa o comportamento de longo prazo da série temporal. Indica se a variável está crescendo, diminuindo ou se mantém relativamente estável ao longo do tempo. Quando você olha para um dado econômico, apesar de haver flutuações de curto prazo, claramente há uma tendência de **longo prazo** na série inteira. Um exemplo é o PIB, apesar de haver flutuações, em geral há claramente uma tendência de crescimento.

### **Sazonalidade**

A sazonalidade refere-se a padrões que se repetem em intervalos fixos e previsíveis de tempo, geralmente associados a fatores sazonais, como estações do ano ou períodos do calendário. Um exemplo classíco é a black friday. Nessa epóca do ano as compras no comércio se elevam muito acima da média em relação a outros períodos (perceba que este é um padrão fixo e prevísivel de acontecer, todo ano sabemos que haverá a black friday). 

Em macroeconomia, é fundamental termos a ideia de sazonalidade. Em um exemplo hipotética, se observarmos que as vendas de panettone aumentaram 25% em dezembro de 2024 em relação a novembro de 2024. Será que esse aumento foi decorrente de um aumento de uma atividade econômica mais favorável que aumenta o consumo dos indivíduos ou foi simplesmente pela chegada do natal?

Em geral, quando estamos trabalhando com dados econômicos, ouviremos muito sobre dados dessazonalizados (ou com ajuste sazonal). Esses dados nada mais são do que a série original livre do componente de sazonalidade. Veremos mais como fazer esse ajuste adiante.

### **Ciclo**

O ciclo representa flutuações de médio e longo prazo que ocorrem de forma não fixa, geralmente associadas a fatores econômicos ou políticos. Os ciclos econômicos incluem períodos de expansão e recessão. Diferente da sazonalidade, os ciclos não ocorrem em intervalos regulares e previsíveis.

### **Componente Aleatório**

O ruído ou erro aleatório é a parte da série temporal que não pode ser explicada por tendência, sazonalidade ou ciclo. Representa flutuações imprevisíveis causadas por eventos inesperados. Uma crise financeira ou um desastre natural pode impactar os preços de ações de maneira abrupta e imprevisível. O ruído pode ser visto como o "resto" da série temporal, isto é, aquilo que sobra depois de removermos a tendência, a sazonalidade e os ciclos.

## Decomposição de Séries Temporais

### **Modelo Aditivo**

$$
    Y(t) = T(t) + S(t) + C(t) + \epsilon(t)
$$

No modelo aditivo, os componentes da série são somados para formar o valor observado em cada período. O modelo aditivo é indicado quando (i) A variação da sazonalidade e do ciclo é constante ao longo do tempo, (ii) A amplitude da sazonalidade não muda conforme o nível da série cresce ou diminui e (iii) O modelo funciona bem quando a variação dos componentes é relativamente estável ao longo do tempo.

**Exemplo**: 

Temperatura diária em uma cidade: Se a temperatura média varia ao longo do ano com flutuações relativamente constantes (por exemplo, sempre 5°C a mais no verão e 5°C a menos no inverno), um modelo aditivo pode ser adequado.

### **Modelo Multiplicativo**

$$
    Y(t) = T(t) \cdot S(t) \cdot C(t) \cdot \epsilon(t)
$$

No modelo multiplicativo, os componentes interagem de forma proporcional ao nível da série. O modelo multiplicativo é indicado quando (i) A variação da sazonalidade e do ciclo não é constante, mas proporcional ao nível da série, (ii) A amplitude da sazonalidade aumenta ou diminui conforme a tendência cresce ou decresce e (iii) O modelo é mais adequado para séries temporais com variações sazonais intensas em valores elevados e menores em valores baixos.

**Exemplo**: 

Faturamento de uma empresa ao longo dos anos: Se a empresa cresce, a sazonalidade também cresce em valor absoluto (por exemplo, vendas dobram no final do ano, mas a variação absoluta de vendas é maior quando o faturamento já é alto), um modelo multiplicativo é mais apropriado.


## **O que é um modelo preditivo?**

Um modelo preditivo é um tipo de modelo de machine learning que tem como objetivo principal fazer previsões sobre novos dados com base em padrões aprendidos a partir de dados históricos.

---

Esses modelos utilizam padrões extraídos dos dados para fazer previsões sobre novas observações. Em termos formais, esses modelos buscam aprender uma função $$f: x_{i} \in X \rightarrow Y$$ que relacione os vetores de características $x_{i} \in X$ com a variável alvo $Y$, de forma que seja possível estimar um $Y$ para um novo $x_{i} \in X$ não observado anteriormente.

---

Os modelos preditivos podem ser utilizados para diferentes tipos de tarefas, como prever se um cliente pagará um empréstimo (classificação), estimar o valor de um imóvel (regressão) ou antecipar o comportamento de uma série temporal (como preços de ativos financeiros).

No entanto, as abordagens para construir esse tipo de modelo variam bastante, dependendo da forma como os dados são tratados e da estratégia utilizada para ajustar os parâmetros do modelo. Podemos dividir os modelos preditivos em algumas categorias principais, são elas:



### **1. Modelos Estatísticos e Probabilísticos**

Esses modelos assumem que os dados seguem uma distribuição probabilística específica e ajustam seus parâmetros para maximizar a verossimilhança dos dados observados.

✅ **Objetivo:** Estimar a distribuição dos dados e inferir a relação entre as variáveis.

**Exemplos:**

- Regressão Linear: Assume que a relação entre as variáveis é linear e que os resíduos seguem uma distribuição normal.

- Regressão Logística: Usa a função sigmoide para modelar probabilidades em problemas de classificação binária.

- Naïve Bayes: Baseia-se na regra de Bayes e assume independência condicional entre os atributos.

📌 **Diferencial:** Esses modelos são baseados em conceitos estatísticos bem definidos e geralmente interpretáveis.

### **2. Modelos Baseados em Otimização**

Esses modelos aprendem a função $f$ otimizando uma métrica específica, sem necessariamente assumir uma distribuição subjacente dos dados.

✅ **Objetivo:** Encontrar um modelo que minimize uma função de erro ou maximize uma margem de separação.

**Exemplos:**

- SVM (Máquinas de Vetores de Suporte): Maximiza a margem entre as classes em problemas de classificação.

- Redes Neurais (MLPs clássicas): Ajustam pesos para minimizar um erro através do gradiente descendente.

📌 **Diferencial:** Baseiam-se em técnicas numéricas de otimização e não exigem uma suposição explícita sobre a distribuição dos dados.

### **3. Modelos Baseados em Particionamento de Dados**

Esses modelos criam regras para dividir os dados em subconjuntos homogêneos, organizando-os hierarquicamente.

✅ **Objetivo:** Criar regras de decisão que segmentem os dados em diferentes categorias.

**Exemplos:**

- Árvores de Decisão: Criam um conjunto de regras sequenciais que levam a decisões.

- Random Forests: Conjunto de árvores de decisão para aumentar a robustez.

📌 **Diferencial:** São fáceis de interpretar e não exigem transformações matemáticas complexas.

### **4. Modelos Baseados em Agrupamento (Clustering)**

Modelos que tentam encontrar padrões e estruturas nos dados sem rótulos supervisionados.

✅ **Objetivo:** Agrupar os dados de forma a maximizar a semelhança dentro dos clusters e minimizar entre os clusters.

**Exemplos:**

- K-Means: Baseado na minimização da soma das distâncias dos pontos ao centro do cluster.

- DBSCAN: Detecta clusters com base na densidade dos dados.

📌 **Diferencial:** Úteis para análise exploratória e segmentação de dados sem supervisão.



### **5. Modelos Baseados em Instâncias**

Esses modelos fazem previsões baseadas nos exemplos mais próximos no conjunto de treinamento, sem construir explicitamente uma função generalizada durante o aprendizado.

✅ **Objetivo:** Utilizar diretamente os dados observados para fazer previsões sem uma fase explícita de aprendizado de parâmetros.

**Exemplo:**
    
- KNN (K-Nearest Neighbors): Classifica ou estima valores com base nos $k$ vizinhos mais próximos.

📌 Diferencial: São conhecidos como modelos preguiçosos (lazy learners) porque não possuem uma fase explícita de treinamento; todo o esforço computacional ocorre no momento da previsão.

### **A importância da amostra de treinamento para um modelo preditivo**

No fim das contas, modelos preditivos tentam encontrar uma função matemática que capture a relação entre as variáveis de entrada (características) e a variável de saída (o que queremos prever). Essa função pode ser vista como uma aproximação da verdadeira relação existente nos dados. No entanto, os dados usados para treinar o modelo representam apenas uma amostra de um conjunto muito maior e desconhecido. Se essa amostra for pequena, enviesada ou não representativa, o modelo pode aprender padrões artificiais que não se repetem em novos dados — um problema conhecido como overfitting.

Diferentes modelos fazem suposições distintas sobre essa relação. Modelos estatísticos assumem que os dados seguem certas distribuições probabilísticas, enquanto modelos baseados em otimização simplesmente ajustam funções para melhor se encaixar nos exemplos observados, sem pressupor uma estrutura fixa. Em ambos os casos, a qualidade da amostra influencia diretamente a capacidade do modelo de fazer boas previsões.

Por isso, ao treinar um modelo, é essencial garantir que os dados de entrada representem bem o fenômeno que estamos tentando modelar. Métodos como validação cruzada, regularização e aumento da diversidade dos dados ajudam a reduzir vieses e melhorar a capacidade do modelo de generalizar para novas situações.
