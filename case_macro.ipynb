# **2. Séries Temporais**

Uma série temporal é um conjunto de observações de uma variável ordenadas no tempo. Essas observações são coletadas em intervalos regulares, como dias, meses, trimestres ou anos, permitindo analisar a evolução dessa variável ao longo do tempo. Diferentemente de quando trabalhamos com dados do tipo cross-section (uma amostra de dados coletadas em um momento específico do tempo), as séries temporais possuem uma **ordenação natural** que é o tempo. Cada observação está associada a um momento específico do tempo e carrega informação sobre a evolução da série.

É fundamental termos em mente alguns conceitos de séries temporais quando estamos trabalhando com macroeconomia. Quando pensamos em juros, inflação, balança comercial, entre outros dados econômicos estamos sempre falando de séries temporais. Pense nas notícias de jornais: IPCA fecha o mês de janeiro em x%, PIB do Brasil cresce y% em 2023, etc...

Perceba que todas essas notícias sempre apresentam o período a que o dado se refere. Não faz sentido falarmos só que a inflação foi de 4.5%, é **fundamental** sabermos do período que o dado se refere.

📝 **Observação**: Neste case, traremos alguns tópicos que sejam mais relevantes e que com uma abordagem mais direcionada. Porém, para um estudo mais aprofundado e formal sobre séries temporais, recomendamos ver o **case de finanças quantitativas**

## **Variável aleatória**
De maneira simples, podemos pensar em uma variável aleatória como um número que representa a realização de um determinado evento que possuí alguma incerteza associada. Pense no tempo que você leva para se deslocar ao trabalho. Esse tempo não é fixo, ele na realidade possui alta incerteza devido ao trânsito, chuva, etc. Esse tempo medido é uma variável aleatória.


## **Processos estocásticos**

Um processo estocástico é basicamente um conjunto de variáveis aleatórias organizadas no tempo. Ou seja, é um fenômeno aleatório que evolui ao longo do tempo.

Suponha que agora, você mede o tempo de deslocamento ao trabalho todos os dias por um ano inteiro. Você terá uma sequência de números diferentes ao longo do tempo, mas seguindo um certo padrão (por exemplo, pode ser maior em dias de chuva ou menor em feriados). Isso é fundamental pois, apesar de cada dia haver certa incerteza associada, é possível modelar o comportamento do deslocamento ao trabalho e extrair informações extramemente úteis.

💡**Dica**: Dá para pensar em um processo estocástico como uma “história” contada por várias variáveis aleatórias ao longo do tempo.

**Definição formal de Série Temporal**:

Seja $(\Omega, \mathcal{F}, P)$ um espaço de probabilidade e seja $\{X_{t}\}_{t \in T}$ um processo estocástico, ou seja, uma família de variáveis aleatórias $X_{t}: \Omega \rightarrow \mathbb{R}$ indexadas por um conjunto $T$, que representa o tempo.

**Uma série temporal é uma realização desse processo estocástico**, isto é, uma sequência de valores observados:

$$
        \{x_{t}\}_{t \in T} \subset \mathbb{R}
$$

onde cada $x_{t}$ corresponde ao valor assumido pelo processo no instante $t$.

## **Componentes de Séries Temporais**

Os componentes de uma série temporal são os diferentes elementos que influenciam a variação dos dados ao longo do tempo. Eles ajudam a decompor uma série temporal em partes distintas para melhor análise e modelagem. Quando pensamos nos componentes das séries temporais, podemos pensar em questões do tipo: Oque está acontecendo com essa série no longo prazo? Existe algum padrão que se repete em intervalos regulares de tempo? Existe algo imprevisível que acontece com essa série?

Ao decompormos uma série temporal, temos uma maior capacidade para responder questões desse tipo. Os principais componentes são:

### **Tendência**

A tendência representa o comportamento de longo prazo da série temporal. Indica se a variável está crescendo, diminuindo ou se mantém relativamente estável ao longo do tempo. Quando você olha para um dado econômico, apesar de haver flutuações de curto prazo, claramente há uma tendência de **longo prazo** na série inteira. Um exemplo é o PIB, apesar de haver flutuações, em geral há claramente uma tendência de crescimento.

### **Sazonalidade**

A sazonalidade refere-se a padrões que se repetem em intervalos fixos e previsíveis de tempo, geralmente associados a fatores sazonais, como estações do ano ou períodos do calendário. Um exemplo classíco é a black friday. Nessa epóca do ano as compras no comércio se elevam muito acima da média em relação a outros períodos (perceba que este é um padrão fixo e prevísivel de acontecer, todo ano sabemos que haverá a black friday). 

Em macroeconomia, é fundamental termos a ideia de sazonalidade. Em um exemplo hipotética, se observarmos que as vendas de panettone aumentaram 25% em dezembro de 2024 em relação a novembro de 2024. Será que esse aumento foi decorrente de um aumento de uma atividade econômica mais favorável que aumenta o consumo dos indivíduos ou foi simplesmente pela chegada do natal?

Em geral, quando estamos trabalhando com dados econômicos, ouviremos muito sobre dados dessazonalizados (ou com ajuste sazonal). Esses dados nada mais são do que a série original livre do componente de sazonalidade. Veremos mais como fazer esse ajuste adiante.

### **Ciclo**

O ciclo representa flutuações de médio e longo prazo que ocorrem de forma não fixa, geralmente associadas a fatores econômicos ou políticos. Os ciclos econômicos incluem períodos de expansão e recessão. Diferente da sazonalidade, os ciclos não ocorrem em intervalos regulares e previsíveis.

### **Componente Aleatório**

O ruído ou erro aleatório é a parte da série temporal que não pode ser explicada por tendência, sazonalidade ou ciclo. Representa flutuações imprevisíveis causadas por eventos inesperados. Uma crise financeira ou um desastre natural pode impactar os preços de ações de maneira abrupta e imprevisível. O ruído pode ser visto como o "resto" da série temporal, isto é, aquilo que sobra depois de removermos a tendência, a sazonalidade e os ciclos.

## **Decomposição de Séries Temporais**

### **Modelo Aditivo**

$$
    Y(t) = T(t) + S(t) + C(t) + \epsilon(t)
$$

No modelo aditivo, os componentes da série são somados para formar o valor observado em cada período. O modelo aditivo é indicado quando (i) A variação da sazonalidade e do ciclo é constante ao longo do tempo, (ii) A amplitude da sazonalidade não muda conforme o nível da série cresce ou diminui e (iii) O modelo funciona bem quando a variação dos componentes é relativamente estável ao longo do tempo.

**Exemplo**: 

Temperatura diária em uma cidade: Se a temperatura média varia ao longo do ano com flutuações relativamente constantes (por exemplo, sempre 5°C a mais no verão e 5°C a menos no inverno), um modelo aditivo pode ser adequado.

### **Modelo Multiplicativo**

$$
    Y(t) = T(t) \cdot S(t) \cdot C(t) \cdot \epsilon(t)
$$

No modelo multiplicativo, os componentes interagem de forma proporcional ao nível da série. O modelo multiplicativo é indicado quando (i) A variação da sazonalidade e do ciclo não é constante, mas proporcional ao nível da série, (ii) A amplitude da sazonalidade aumenta ou diminui conforme a tendência cresce ou decresce e (iii) O modelo é mais adequado para séries temporais com variações sazonais intensas em valores elevados e menores em valores baixos.

**Exemplo**: 

Faturamento de uma empresa ao longo dos anos: Se a empresa cresce, a sazonalidade também cresce em valor absoluto (por exemplo, vendas dobram no final do ano, mas a variação absoluta de vendas é maior quando o faturamento já é alto), um modelo multiplicativo é mais apropriado.

## **Como estimar os componentes de uma série temporal?**

### **Médias móveis**

A estimativa da sazonalidade usando um fator de correção por médias móveis é um método simples e eficaz para identificar padrões sazonais em séries temporais. Esse método consiste em remover a tendência da série através da suavização dos dados com médias móveis e, em seguida, calcular os fatores sazonais.

**Média Móvel centrada**

- janela de pontos é centralizada no ponto atual, incluindo pontos anteriores e posteriores. 

- Por exemplo, se a janela tem tamanho 5, a média no ponto $t$ será calculada usando os pontos $t_{-2}$, $t_{-1}$, $t$, $t_{+1}$, $t_{+2}$

Utilizamos a média móvel para estimar a tendência da série sem ajuste sazonal. Iremos usar essa estimação da têndencia para remover o compontente sazonal.

A fórmula para a média móvel centrada é a seguinte:

$$
\tilde{X_t} = \frac{1}{h} \sum_{s=t - \frac{(h-1)}{2}}^{t + \frac{(h-1)}{2}} X_s
$$

Feito o cálculo da média móvel centrada, podemos calcular um fator de correção sazonal:

$$
\hat{\delta}_s = \frac{1}{\lfloor (T - s)/h \rfloor + 1} \sum_{t=s, s+h, s+2h, \dots} (X_t - \tilde{X_t})
$$

Obs: Não se assuta com essa fórmula, oque ela faz é basicamente calcular a média dos desvios entre os valores reais ($X_t$) e a tendência estimada $\tilde{X_t}$ para um determinado período sazonal $s$ (exemplo: todos os meses de janeiro em uma série mensal). 

O fator de correção sazonal nos dá um valor que representa o quanto um determinado período tende a se desviar da tendência geral. Se os fatores forem positivos, indicam que nesse período os valores costumam ser maiores que a tendência. Se forem negativos, indicam que nesse período os valores costumam ser menores que a tendência. Esse fator pode então ser removido da série para obter uma versão "dessazonalizada" dos dados, permitindo análise mais clara de tendências e ciclos.

Por fim, temos o fator recentrado:

$$
\tilde{\delta}_s = \hat{\delta}_s - \frac{1}{h} \sum_{j=1}^{h} \hat{\delta}_j
$$

Recentramos os fatores sazonais para que sua média seja exatamente zero (modelo aditivo) ou 1 (modelo multiplicativo). Isso evita distorções nos dados corrigidos.

###  **X-13 ARIMA-SEATS**

O X-13-ARIMA-SEATS é um método estatístico desenvolvido pelo U.S. Census Bureau para realizar ajustes sazonais e análises de séries temporais. Ele é amplamente utilizado por economistas, estatísticos e governos para remover efeitos sazonais e entender melhor a tendência e o ciclo econômico das séries temporais.

Combina metodologia de médias móveis com a possibilidade de inclusão de efeitos-calendário (por padrão, modelo controla por dias úteis do mês e alguns feriados móveis; passível de alteração), detecção automática de outliers, backfitting para completamento da série de tempo e seleção automática de modelo aditivo ou multiplicativo.

### **Separando Ciclo da Tendência**

Supondo agora que você possua uma série livre do componente sazonal (dessazonalizada). Como separar a tendência do ciclo de uma série?

**Filtro HP**

O filtro de Hodrick-Prescott (HP) é um método estatístico usado para separar a tendência de longo prazo de uma série temporal das flutuações cíclicas e sazonais. Ele é amplamente utilizado em macroeconomia para analisar o ciclo econômico e identificar o crescimento potencial de variáveis como o PIB, a inflação e o desemprego.

De maneira simplificada, o filtro HP busca estimar uma têndencia que minimiza uma função de erro. Quando estamos trabalhando com ele, devemos utilizar um valor $\lambda$ que controla o grau de suavização da tendência. Os valores mais comuns são:

- $\lambda = 1600$ para séries trimestrais (exemplo: PIB trimestral).
- $\lambda = 6.25$ para séries anuais.
- $\lambda = 129600$ para séries mensais.

**Instabilidade na ponta do filtro HP**

A instabilidade na ponta do filtro HP refere-se ao fato de que a tendência estimada nos períodos mais recentes (ou seja, no final da série) pode ser imprecisa e altamente sensível a novos dados. Isso ocorre porque o filtro HP é simétrico, ou seja, ele usa informações dos períodos anteriores e posteriores para calcular a tendência. No entanto, nos últimos pontos da série, não há valores futuros disponíveis, o que enfraquece a suavização e pode levar a oscilações artificiais na tendência estimada.

Esse problema faz com que a tendência no final da série seja irrealista, sofrente várias alterações frequentementes a medida que novos dados vão sendo adicionados.

**Método de Hamilton**

Em vez de usar um suavizador estatístico fixo como o filtro HP, o método de Hamilton utiliza uma regressão autorregressiva (AR) para estimar a tendência.

$$
T_t = \beta_0 + \beta_1 Y_{t-h} + \beta_2 Y_{t-2h} + ... + \beta_p Y_{t-ph} + \varepsilon_t
$$

Onde:

- $T_t$ é a tendência estimada no tempo $ t $.
- $ Y_{t-h}, Y_{t-2h}, ..., Y_{t-ph} $ são os valores passados da série temporal, utilizados como preditores.
- $ h $ é o horizonte de suavização (por exemplo, $ h = 8 $ para uma série trimestral).
- $ p $ é o número de defasagens incluídas no modelo (Hamilton sugere $ p = 4 $).
- $ \beta_0, \beta_1, \beta_2, ... $ são os coeficientes da regressão, ajustados para minimizar o erro.
- $ \varepsilon_t $ é o erro residual da regressão.

Ele é mais robusto, não sofre de instabilidade na ponta e tem uma melhor interpretação econômica. Amplamente utilizado em análises macroeconômicas, como PIB potencial e ciclos econômicos.

📝**Observação**:

Não se preocupe em entender tudo sobre componentes de séries temporais. A ideia aqui é que você tenha em mente os fundamentos ciêntificos por trás de coisas que utilizamos no dia a dia em séries econômicas

### **Método Clássico de Decomposição de Séries Temporais**

O método clássico de decomposição é uma abordagem estatística usada para separar uma série temporal em seus principais componentes. Na prática, ele pode ser uma boa primeira abordagem para decomposição visto sua rápida e facil implementação. O método clássico de decomposição ajuda a separar os diferentes padrões de uma série temporal, tornando mais fácil entender o que está acontecendo nos dados ao longo do tempo. Ele divide a série em tendência, sazonalidade, ciclo e ruído.

O passo a passo do método clássico é o seguinte:

1. Encontrar a Tendência $(T_t)$ → O caminho de longo prazo

Para encontrá-la, usamos uma média móvel, que basicamente tira a média dos valores ao redor de um ponto para suavizar as oscilações.

2. Descobrir a Sazonalidade $(S_t)$ → Os padrões que se repetem

O objetivo aqui é identificar essas oscilações regulares e entender como elas afetam os dados. Para isso, pegamos os valores de um mesmo período ao longo dos anos (por exemplo, todas as vendas de dezembro) e calculamos uma média para descobrir o quanto esse mês tende a ser maior ou menor do que o normal.

3. Identificar o Ciclo $(C_t)$ → Grandes oscilações que não seguem um padrão fixo

Para encontrar o ciclo, removemos a tendência e a sazonalidade da série e observamos as variações que sobram.

4. Separar o Ruído $(e_t)$ → As variações aleatórias

Por fim, depois de remover a tendência, a sazonalidade e o ciclo, ainda sobram pequenas variações que não seguem nenhum padrão aparente. Isso é o ruído, ou seja, as oscilações causadas por fatores imprevisíveis, como um feriado inesperado, uma greve ou um erro de medição.

Esse método é facilmente implementado em python utilizando a biblioteca *statsmodels*

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

# Gerando uma série temporal fictícia com tendência, sazonalidade e ruído
np.random.seed(42)
time = pd.date_range(start='2015-01-01', periods=60, freq='M')  # 60 meses (5 anos)
trend = np.linspace(50, 100, 60)  # Tendência linear crescente
seasonality = 10 * np.sin(2 * np.pi * np.arange(60) / 12)  # Padrão sazonal anual
noise = np.random.normal(0, 3, 60)  # Ruído aleatório
data = trend + seasonality + noise  # Série final combinada

# Criando um DataFrame
df = pd.DataFrame({'Date': time, 'Value': data})
df.set_index('Date', inplace=True)

# Aplicando a decomposição sazonal aditiva
decomposition = sm.tsa.seasonal_decompose(df['Value'], model='additive', period=12)

# Criando o gráfico dos componentes da série temporal
fig, axes = plt.subplots(4, 1, figsize=(10, 8), sharex=True)

axes[0].plot(df['Value'], label='Série Original')
axes[0].set_title('Série Temporal Original')
axes[0].legend()

axes[1].plot(decomposition.trend, label='Tendência', color='tab:orange')
axes[1].set_title('Tendência')
axes[1].legend()

axes[2].plot(decomposition.seasonal, label='Sazonalidade', color='tab:green')
axes[2].set_title('Sazonalidade')
axes[2].legend()

axes[3].plot(decomposition.resid, label='Resíduo (Ruído)', color='tab:red')
axes[3].set_title('Ruído')
axes[3].legend()

plt.xlabel('Data')
plt.tight_layout()
plt.show()
```

## **Como estacionarizar Séries Temporais?**

Vimos anteriormente a importância de trabalharmos com séries estacionárias quando pensamos em séries temporais. Dessa forma, quando formos pensar em uma modelagem de série econômica precisamos pensar se a série que temos é estacionário e, em caso negativo, como **estacionarizar-la**

**Processos integrados**

De maneira simplificada, um **processo integrado de ordem d ou ($I(d)$)** é uma série temporal que não é estacionária, mas pode se tornar estacionária ao aplicar diferenciações sucessivas. Em outras palavras, uma série integrada precisa ser transformada para remover tendência ou dependências de longo prazo antes de ser analisada ou modelada. 

A ordem d indica quantas diferenciações são necessárias para estacionarizar a série em questão. Exemplo:

- Série $I(0)$, indica uma série que é estacionária (não precisa de nenhuma diferenciação)
- Série $I(1)$, indica uma série que precisa de uma diferenciação para ser estacionária.
- Série $I(2)$, indica uma série que precisa de duas diferenciações para ser estacionária.

**Qual a importância de estacionariedade?**

Processos $I(d)$ com $d > 0$ geram graves problemas de inferência (regressão espúria). Ou seja, se regredirmos duas séries que são $I(d)$ com $d > 0$ podemos acabar tendo um caso em que a regressão indica que as duas séries possuem relação quando na verdade não há.

Dessa forma, é importante que possamos identificar se nossa série possui uma tendência estocástica (ou seja, é um processo não estacionário) e fazer as devidas transformações.

📝**Observação**:

A maior parte das séries econômicas são do tipo $I(1)$ possuem tendência estocásticas requerendo apenas uma diferenciação para se tornar estacionária. Dessa forma, daremos enfase a esses casos. Porém, tenha em mente que oque for coberto aqui é apenas uma pequena parte da área de estudo em Séries Temporais. Caso se interesse, sugerimos procurar outras fontes para um maior aprofundamento.

### **Teste de Raiz Unitária**

O teste de raiz unitária é um método estatístico usado para verificar se uma série temporal é estacionária ou não estacionária. Ele ajuda a identificar se os dados possuem uma tendência persistente ou se os valores oscilam em torno de uma média constante.

Imagine que você está observando a temperatura média de uma cidade ao longo dos anos. Se a temperatura tem um padrão estável, variando sempre em torno de uma média fixa (como 25°C), dizemos que essa série é estacionária. Agora, se a temperatura sobe continuamente devido ao aquecimento global, então há uma tendência de longo prazo, e a série é não estacionária.

O teste de raiz unitária nos ajuda a matematicamente decidir se uma série se comporta como a primeira situação (estacionária) ou a segunda (não estacionária).

- Considere o seguinte modelo para uma série de tempo:

$$
Y_t = \rho Y_{t-1} + \varepsilon_t
$$

Onde:
- $Y_t$ → Valor atual da série no tempo $t$.
- $\rho$ → Coeficiente de persistência (quão forte a influência do passado no presente).
- $\varepsilon_t$ → Erro aleatório.

Observe que:
1.  Se $|\rho| < 1$ , processo é $I(0)$
2.  Se $\rho = 1$, processo é $I(1)$

Subtraindo $Y_{t-1}$ de ambos os lados temos:

$$
\Delta Y_t = \gamma Y_{t-1} + \varepsilon_t
$$

Observe que:
1.  Se $\gamma = 0$ , processo é $I(1)$
2.  Se $\gamma \in (0, 2)$, processo é $I(0)$

Assim, num teste de raiz unitária queremos saber se $\gamma$ é igual a 0. Se for, a série possui **uma raiz unitária e não é estacionária.**

**Hipóteses do Teste de Raiz Unitária:**
- **Hipótese nula** ($H_0$): A série tem uma raiz unitária ($\gamma = 0$) → **Não estacionária**
- **Hipótese alternativa** ($H_1$): A série não tem raiz unitária ($\gamma = 1$) → **Estacionária**

**Interpretação dos Resultados**

- Se o valor-p do teste for **maior que 0.05**, não rejeitamos $H_0$→ a série tem raiz unitária (não estacionária).
- Se o valor-p for **menor** que 0,05, rejeitamos $H_0$ → a série é estacionária.

Devido ao fato de, sob a nula, o processo apresentar tendência estocástica, a distribuição de referência da estatística do teste não é normal. Dessa maneira, é necessário utilizar valores críticos tabulados (Dickey e Fuller, 1979).

📝**Observação**: O teste de raiz unitária mais utilizado na prática é o teste ADF. Porém, existe um procedimento muito mais extenso para fazer uma checagem mais robusta. Caso tenha interesse, recomendamos procurar outras fontes para se aprofundar no assunto.

Abaixo um código em python de como aplicaríamos esse teste.

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
import matplotlib.pyplot as plt

# Criar um passeio aleatório (Random Walk)
np.random.seed(42)
n = 200
y = np.cumsum(np.random.normal(0, 1, n))  # Passeio aleatório (não estacionário)

# Aplicar o teste ADF
adf_test = adfuller(y)
print(f'Valor-p do Teste ADF: {adf_test[1]:.5f}')

# Plotar a série
plt.figure(figsize=(8,4))
plt.plot(y, label="Passeio Aleatório")
plt.title("Teste de Raiz Unitária - Dickey-Fuller Aumentado")
plt.legend()
plt.show()
```

#### **O que fazer caso a série tenha raiz unitária? (ou seja, não é estacionária)**
- Caso a série apresente raiz unitária, trabalharemos com ela na primeira diferença
- Caso a série não possua raiz unitária, podemos trabalhar com ela em nível de forma normal


## **Modelos Univariados**

Os modelos univariados de séries temporais são usados para modelar a dependência temporal de uma série e, dessa forma, fazer previsões de seus valores futuros. Os modelos mais utilizados são: **AR (AutoRegressive)**, **MA (Moving Average)** e **ARMA (AutoRegressive Moving Average)**

### **Modelo Autorregressivo (AR)**

O modelo **$AR(p)$** descreve uma série temporal onde o valor atual depende linearmente de seus valores passados. A ordem $p$ indica quantas defasagens devem ser incluídas na modelagem. Podemos pensar na ordem $p$ como o número de defasagens da série que ainda trazem alguma informação relevante para o modelo. Pense em uma série de inflação, intuitivamente devemos pensar que a inflação do mês passado carrega bastante informação para a inflação do mês atual. Porém, é difícil imaginar que a inflação de março de 1996 seja muito relevante para prever a inflação de janeiro de 2025.

Formalmente, podemos definir um modelo $AR$ pela seguinte equação:

$$
Y_t = \phi_1 Y_{t-1} + \phi_2 Y_{t-2} + ... + \phi_p Y_{t-p} + \varepsilon_t
$$
Onde:
- $Y_t$ é o valor da série no tempo $t$
- $Y_{t-1}$, $Y_{t-2}$,..., $Y_{t-p}$ são os valores passados da série (lags).
- $\phi_1$, $\phi_2$,...,$\phi_p$ são os coeficientes do modelo que indicam a influência dos valores passados.
- $\varepsilon_t$ é o erro aleatório com média zero e variância constante.

**Exemplo Prático:**

Imagine que você está prevendo a temperatura de hoje com base na temperatura dos últimos dias. Se a temperatura de hoje for parecida com a dos dias anteriores, isso sugere que um modelo AR pode ser adequado.

O código abaixo fornece um exemplo de como implementar um modelo AR em Python

```python
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima_process import ArmaProcess

# Definir semente para reprodutibilidade
np.random.seed(42)
n = 200  # Número de observações

# Parâmetros AR(1) (Autorregressivo)
ar_params = np.array([1, -0.7])  # Coeficiente AR(1) = 0.7
ma_params = np.array([1])  # Sem componente MA

# Criar processo AR(1)
ar_process = ArmaProcess(ar_params, ma_params)
ar_series = ar_process.generate_sample(nsample=n)

# Plotar a série gerada
plt.figure(figsize=(8, 4))
plt.plot(ar_series, label="Modelo AR(1)", color="blue")
plt.title("Série Simulada - AR(1)")
plt.legend()
plt.show()
```

### **Modelo de Médias Móveis (MA)**

O modelo $MA(q)$ modela uma série temporal onde o valor atual depende dos **erros passados (ruído branco)**, em vez dos valores da própria série. Pense nos erros passados como os choques **não antecipados** que afetaram a série em períodos anteriores. Pensando novamente no caso de uma série de inflação, um choque exógeno pode ser uma seca que destrói uma grande safra. Esse choque faz com que a inflação de alimentos aumente não só no mês em questão, mas o efeito do choque pode demorar a se dissipar, afetando períodos posteriores.

Formalmente, podemos definir um modelo $MA$ pela seguinte equação:

$$
Y_t = \varepsilon_t + \theta_1 \varepsilon_{t-1} + \theta_2 \varepsilon_{t-2} + ... + \theta_q \varepsilon_{t-q}
$$

Onde:
- $Y_t$ é o valor da série no tempo $t$
- $\varepsilon_t$ é o erro aleatório com média zero e variância constante.
- $\varepsilon_{t-1}$, $\varepsilon_{t-2}$,...,$\varepsilon_{t-q}$ são os erros anteriores.
- $\theta_1$, $\theta_2$,..., $\theta_q$ são coeficientes que indicam a influência dos erros passados.

**Exemplo Prático:**

Imagine que você está prevendo o preço de um ativo financeiro. Se um evento inesperado influenciou o preço no dia anterior, isso pode ter impacto hoje. O modelo MA captura esse efeito, considerando choques passados no sistema.

O codigo abaixo fornece um exemplo de um modelo MA feito em Python

```python
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima_process import ArmaProcess

# Definir semente para reprodutibilidade
np.random.seed(42)
n = 200  # Número de observações

# Parâmetros MA(1) (Médias Móveis)
ar_params = np.array([1])  # Sem componente AR
ma_params = np.array([1, 0.5])  # Coeficiente MA(1) = 0.5

# Criar processo MA(1)
ma_process = ArmaProcess(ar_params, ma_params)
ma_series = ma_process.generate_sample(nsample=n)

# Plotar a série gerada
plt.figure(figsize=(8, 4))
plt.plot(ma_series, label="Modelo MA(1)", color="green")
plt.title("Série Simulada - MA(1)")
plt.legend()
plt.show()
```

### **Modelo ARMA (Autorregressivo + Médias Móveis)**

O modelo $ARMA$, como o próprio nome já indica, combina a modelagem tanto do $AR$ quanto do $MA$. Ou seja, permite que a série temporal seja modelada tanto em termos de valores passados da série quanto dos erros passados.

```python
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima_process import ArmaProcess

# Definir semente para reprodutibilidade
np.random.seed(42)
n = 200  # Número de observações

# Parâmetros ARMA(1,1) (Autorregressivo + Médias Móveis)
ar_params = np.array([1, -0.5])  # Coeficiente AR(1) = 0.5
ma_params = np.array([1, 0.4])  # Coeficiente MA(1) = 0.4

# Criar processo ARMA(1,1)
arma_process = ArmaProcess(ar_params, ma_params)
arma_series = arma_process.generate_sample(nsample=n)

# Plotar a série gerada
plt.figure(figsize=(8, 4))
plt.plot(arma_series, label="Modelo ARMA(1,1)", color="red")
plt.title("Série Simulada - ARMA(1,1)")
plt.legend()
plt.show()
```

## **O que é um modelo preditivo?**

Um modelo preditivo é um tipo de modelo de machine learning que tem como objetivo principal fazer previsões sobre novos dados com base em padrões aprendidos a partir de dados históricos.

---

Esses modelos utilizam padrões extraídos dos dados para fazer previsões sobre novas observações. Em termos formais, esses modelos buscam aprender uma função $$f: x_{i} \in X \rightarrow Y$$ que relacione os vetores de características $x_{i} \in X$ com a variável alvo $Y$, de forma que seja possível estimar um $Y$ para um novo $x_{i} \in X$ não observado anteriormente.

---

Os modelos preditivos podem ser utilizados para diferentes tipos de tarefas, como prever se um cliente pagará um empréstimo (classificação), estimar o valor de um imóvel (regressão) ou antecipar o comportamento de uma série temporal (como preços de ativos financeiros).

No entanto, as abordagens para construir esse tipo de modelo variam bastante, dependendo da forma como os dados são tratados e da estratégia utilizada para ajustar os parâmetros do modelo. Podemos dividir os modelos preditivos em algumas categorias principais, são elas:



### **1. Modelos Estatísticos e Probabilísticos**

Esses modelos assumem que os dados seguem uma distribuição probabilística específica e ajustam seus parâmetros para maximizar a verossimilhança dos dados observados.

✅ **Objetivo:** Estimar a distribuição dos dados e inferir a relação entre as variáveis.

**Exemplos:**

- Regressão Linear: Assume que a relação entre as variáveis é linear e que os resíduos seguem uma distribuição normal.

- Regressão Logística: Usa a função sigmoide para modelar probabilidades em problemas de classificação binária.

- Naïve Bayes: Baseia-se na regra de Bayes e assume independência condicional entre os atributos.

📌 **Diferencial:** Esses modelos são baseados em conceitos estatísticos bem definidos e geralmente interpretáveis.

### **2. Modelos Baseados em Otimização**

Esses modelos aprendem a função $f$ otimizando uma métrica específica, sem necessariamente assumir uma distribuição subjacente dos dados.

✅ **Objetivo:** Encontrar um modelo que minimize uma função de erro ou maximize uma margem de separação.

**Exemplos:**

- SVM (Máquinas de Vetores de Suporte): Maximiza a margem entre as classes em problemas de classificação.

- Redes Neurais (MLPs clássicas): Ajustam pesos para minimizar um erro através do gradiente descendente.

📌 **Diferencial:** Baseiam-se em técnicas numéricas de otimização e não exigem uma suposição explícita sobre a distribuição dos dados.

### **3. Modelos Baseados em Particionamento de Dados**

Esses modelos criam regras para dividir os dados em subconjuntos homogêneos, organizando-os hierarquicamente.

✅ **Objetivo:** Criar regras de decisão que segmentem os dados em diferentes categorias.

**Exemplos:**

- Árvores de Decisão: Criam um conjunto de regras sequenciais que levam a decisões.

- Random Forests: Conjunto de árvores de decisão para aumentar a robustez.

📌 **Diferencial:** São fáceis de interpretar e não exigem transformações matemáticas complexas.

### **4. Modelos Baseados em Agrupamento (Clustering)**

Modelos que tentam encontrar padrões e estruturas nos dados sem rótulos supervisionados.

✅ **Objetivo:** Agrupar os dados de forma a maximizar a semelhança dentro dos clusters e minimizar entre os clusters.

**Exemplos:**

- K-Means: Baseado na minimização da soma das distâncias dos pontos ao centro do cluster.

- DBSCAN: Detecta clusters com base na densidade dos dados.

📌 **Diferencial:** Úteis para análise exploratória e segmentação de dados sem supervisão.



### **5. Modelos Baseados em Instâncias**

Esses modelos fazem previsões baseadas nos exemplos mais próximos no conjunto de treinamento, sem construir explicitamente uma função generalizada durante o aprendizado.

✅ **Objetivo:** Utilizar diretamente os dados observados para fazer previsões sem uma fase explícita de aprendizado de parâmetros.

**Exemplo:**
    
- KNN (K-Nearest Neighbors): Classifica ou estima valores com base nos $k$ vizinhos mais próximos.

📌 Diferencial: São conhecidos como modelos preguiçosos (lazy learners) porque não possuem uma fase explícita de treinamento; todo o esforço computacional ocorre no momento da previsão.

### **A importância da amostra de treinamento para um modelo preditivo**

No fim das contas, modelos preditivos tentam encontrar uma função matemática que capture a relação entre as variáveis de entrada (características) e a variável de saída (o que queremos prever). Essa função pode ser vista como uma aproximação da verdadeira relação existente nos dados. No entanto, os dados usados para treinar o modelo representam apenas uma amostra de um conjunto muito maior e desconhecido. Se essa amostra for pequena, enviesada ou não representativa, o modelo pode aprender padrões artificiais que não se repetem em novos dados — um problema conhecido como overfitting.

Diferentes modelos fazem suposições distintas sobre essa relação. Modelos estatísticos assumem que os dados seguem certas distribuições probabilísticas, enquanto modelos baseados em otimização simplesmente ajustam funções para melhor se encaixar nos exemplos observados, sem pressupor uma estrutura fixa. Em ambos os casos, a qualidade da amostra influencia diretamente a capacidade do modelo de fazer boas previsões.

Por isso, ao treinar um modelo, é essencial garantir que os dados de entrada representem bem o fenômeno que estamos tentando modelar. Métodos como validação cruzada, regularização e aumento da diversidade dos dados ajudam a reduzir vieses e melhorar a capacidade do modelo de generalizar para novas situações.
